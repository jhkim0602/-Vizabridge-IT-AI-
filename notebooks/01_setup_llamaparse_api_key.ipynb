# LlamaParse API Key Setup

이 노트북은 LlamaParse/LlamaCloud API 키를 로컬 `.env`에서 읽는지 확인합니다.

1. LlamaCloud에서 API Key를 생성합니다: https://cloud.llamaindex.ai
2. 프로젝트 루트에 `.env` 파일을 만들고 아래 값을 입력합니다.

```bash
LLAMA_CLOUD_API_KEY=llx-...
```

API 키는 노트북이나 코드에 직접 적지 않습니다.

````python
from pathlib import Path

def find_project_root(start: Path | None = None) -> Path:
    current = (start or Path.cwd()).resolve()
    for path in [current, *current.parents]:
        if (path / "requirements.txt").exists() and (path / "data" / "raw").exists():
            return path
    raise RuntimeError("프로젝트 루트를 찾지 못했습니다.")

PROJECT_ROOT = find_project_root()
PROJECT_ROOT
````

````python
# 최초 1회만 실행하세요.
%pip install -r {PROJECT_ROOT / 'requirements.txt'}
````

````python
import os
from dotenv import load_dotenv

env_path = PROJECT_ROOT / ".env"

load_dotenv(env_path)
api_key = os.getenv("LLAMA_CLOUD_API_KEY")

if not api_key:
    raise RuntimeError(
        f"LLAMA_CLOUD_API_KEY가 설정되지 않았습니다. {env_path} 파일을 확인해주세요."
    )

print(f"LLAMA_CLOUD_API_KEY loaded: {api_key[:7]}...{api_key[-4:]}")
````

````python
from llama_cloud import LlamaCloud

# LlamaCloud()는 기본적으로 LLAMA_CLOUD_API_KEY 환경변수를 사용합니다.
client = LlamaCloud()
print("LlamaCloud client initialized")
````
